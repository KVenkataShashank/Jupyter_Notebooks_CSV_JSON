```python
required_deal_details = {"deal_id", "deal_type", "deal_title", "deal_photo", "deal_state", "deal_url", 
                         "deal_starts_at", "deal_ends_at", "deal_price", "list_price"}

country_wise_deals_details = {}

for country in new_country_list:
    country_wise_deals_details[country] = []
    num_of_deals = len(processed_country_wise_data[country]["data"]["deals"])
    
    for deal_no in range(num_of_deals):
        deal_details = {}
        deal_price = 0
        list_price = 0
        deal_price_currency = ""
        list_price_currency = ""
        for key, value in processed_country_wise_data[country]["data"]["deals"][deal_no].items():
            if key in required_deal_details:
                if key == "deal_title":
                    deal_details[key] = value.split(",")[0]
                else:
                    deal_details[key] = value

                if key == "deal_price":
                    deal_price = float(value["amount"])
                    deal_price_currency = value["currency"]
                elif key == "list_price":
                    list_price = float(value["amount"])
                    list_price_currency = value["currency"]

        if deal_price_currency in country_currencies_rates:
            deal_price_USD = deal_price/country_currencies[deal_price_currency]
        else:
            deal_price_USD = deal_price

        if list_price_currency in country_currencies_rates:
            list_price_USD = list_price/country_currencies[list_price_currency]
        else:
            list_price_USD = list_price

        if deal_price != 0 and list_price != 0:
            if deal_price < list_price:
                deal_details["profit_or_loss_category"] = "PROFIT"
            else:
                deal_details["profit_or_loss_category"] = "LOSS"

        if deal_price_currency != "":
            deal_details["deal_price_USD"] = deal_price_USD
        if list_price_currency != "":
            deal_details["list_price_USD"] = list_price_USD
        
        country_wise_deals_details[country].append(deal_details)
```
